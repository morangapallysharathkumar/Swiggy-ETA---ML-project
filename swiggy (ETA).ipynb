## handling missing values

```python
X.isnull().sum()
```

```python
# Select numerical columns
num_cols = X.select_dtypes(include=['int64', 'float64']).columns
```

```python
# Fill missing values in numerical columns with median
X[num_cols] = X[num_cols].fillna(X[num_cols].median())
```

```python
# Select categorical columns
cat_cols = X.select_dtypes(include=['object', 'string', 'category']).columns
```

```python
# Fill missing values in categorical columns with mode
for col in cat_cols:
    X[col] = X[col].fillna(X[col].mode()[0])
```

## Removing irrelevent columns

```python
cols_to_drop = [
    'rider_id',
    'order_date',
    'restaurant_latitude',
    'restaurant_longitude',
    'delivery_latitude',
    'delivery_longitude',
    'order_day',
    'order_month'
]


X = X.drop(columns=cols_to_drop, errors='ignore')
```

## Feature segregation

```python
Nominal_cols = [
    'weather',
    'festival',
    'city_type',
    'type_of_order',
    'type_of_vehicle',
    'order_time_of_day',
    'order_day_of_week',
    'city_name'
]

Ordinal_cols = [
    'traffic'
]

Numerical_cols = [
    'age',
    'ratings',
    'vehicle_condition',
    'multiple_deliveries',
    'pickup_time_minutes',
    'order_time_hour',
    'distance',
    'is_weekend',
]
```

```python
# ✅ FORCE CORRECT DATA TYPES (ADD THIS)
for col in Nominal_cols + Ordinal_cols:
    X[col] = X[col].astype(str)

for col in Numerical_cols:
    X[col] = pd.to_numeric(X[col], errors='coerce')
```

## Data Preprocessing

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder, OrdinalEncoder, PowerTransformer
from sklearn.impute import SimpleImputer



# -------------------- Pipelines --------------------
num_pipeline = Pipeline([
    ('Feature scaling', StandardScaler()),
    ('Feature Transformation', PowerTransformer(method='yeo-johnson'))
])

ord_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),  
    ('encoder', OrdinalEncoder(
        categories=[['low','medium','high','jam']],
        handle_unknown='use_encoded_value',
        unknown_value=-1
    ))
])

nom_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='constant', fill_value='missing')),  # fill missing strings
    ('onehot', OneHotEncoder(drop='first', handle_unknown='ignore'))
])

# -------------------- Column Transformer --------------------
Datapreprocessing = ColumnTransformer([
    ('Num', num_pipeline, Numerical_cols),
    ('Ord', ord_pipeline, Ordinal_cols),
    ('Nom', nom_pipeline, Nominal_cols)
])
```

```python
from sklearn.model_selection import RandomizedSearchCV
from xgboost import XGBRegressor
from sklearn.pipeline import Pipeline

# Pipeline
xgb_pipeline = Pipeline([
    ('preprocessing', Datapreprocessing),   # your preprocessor name
    ('model', XGBRegressor(random_state=23))
])

# Parameter Distribution
param_dist = {
    'model__n_estimators': [200, 300],
    'model__learning_rate': [0.01, 0.03, 0.05],
    'model__max_depth': [3, 4, 5],
    'model__subsample': [0.7, 0.8],
    'model__colsample_bytree': [0.7, 0.8],
    'model__gamma': [0.2, 0.3, 0.5],
    'model__min_child_weight': [5, 7, 10],
    'model__reg_alpha': [0, 0.1, 1],
    'model__reg_lambda': [1, 5, 10]
}

# Randomized Search
random_search = RandomizedSearchCV(
    estimator=xgb_pipeline,
    param_distributions=param_dist,
    n_iter=15,              # reduced (safe for laptop)
    cv=5,
    scoring='r2',
    verbose=2,
    n_jobs=2,               # ⚠️ important (avoid crash)
    random_state=42
)

# Fit
random_search.fit(X_train, y_train)

# Best Results
print("Best Parameters:", random_search.best_params_)
print("Best CV R2:", random_search.best_score_)
```

## input

```python
import pandas as pd

def prepare_input(distance, traffic, pickup_time):
    """
    traffic must be one of: 'low', 'medium', 'high', 'jam'
    """
    # ✅ FIX: Validate traffic input - must be string category, not int
    valid_traffic = ['low', 'medium', 'high', 'jam']
    if traffic not in valid_traffic:
        raise ValueError(f"traffic must be one of {valid_traffic}, got '{traffic}'")
    
    return pd.DataFrame([{
        'distance': distance,
        'traffic': traffic,          # ✅ Must be string: 'low','medium','high','jam'
        'pickup_time_minutes': pickup_time,

        # default values
        'age': 30,
        'ratings': 4.2,
        'vehicle_condition': 2,
        'multiple_deliveries': 1,
        'order_time_hour': 20.0,
        'weather': 'sunny',
        'festival': 'No',
        'city_type': 'urban',
        'type_of_order': 'Meal',
        'type_of_vehicle': 'motorcycle',
        'order_time_of_day': 'evening',
        'order_day_of_week': 'friday',
        'is_weekend': 0
    }])
```

```python
# ✅ FIX: Pass traffic as valid string category, NOT as integer string '2'
# Valid options: 'low', 'medium', 'high', 'jam'

model = grid_search.best_estimator_

input_df = prepare_input(5,'high', 15)   # ✅ FIXED: 'high' instead of '2'

# separate numeric and categorical columns
num_cols = X_train.select_dtypes(include=['int64','float64']).columns
cat_cols = X_train.select_dtypes(include=['object']).columns

# fill missing columns with appropriate types
for col in X_train.columns:
    if col not in input_df.columns:
        if col in num_cols:
            input_df[col] = 0
        else:
            # pick most frequent category from training
            input_df[col] = X_train[col].mode()[0]

# reorder columns
input_df = input_df[X_train.columns]

# now predict safely
prediction = model.predict(input_df)
print("Estimated Delivery Time:", round(prediction[0], 2), "minutes")
```
